### Streaming

Most models can stream their output content while it is being generated. By displaying output progressively, streaming signigicantly improves user experience, particularly for longer responses. Calling Stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time:

```python
# invoke gives the whole output onces it is done generating it
#streaming gives the output while it is being generated (like other LLMs ) 
```

```python
#invoke
model.invoke("why do dogs bark?") #gives the whole output after we wait for its generation
# straiming
for chunk in model.stream("why do dogs bark?"):
    print(chunk.text, flush=True) #flush= true ->Don't keep this in the buffer. Send it to the screen immediately
```

### Batch

Batching a collection of independent requests to a model can siginficantly improve performance and reduce costs, as the processing can be done in prarallel:

```python
### to send multiple questions/prompts to the model parallely and it will give output parallely
model.batch([
    "why do we cry?",
    "how does a bird fly",
],
config={ 'max_concurrency': 5,#limits to 5 parallel calls
}
)
```
